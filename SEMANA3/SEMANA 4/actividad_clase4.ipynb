```python
ventas = [100, 200, 150], [80, 90, 120 ]

print("Total de ventas por vendedor")
for i in range(len(ventas)):
    total = sum(ventas[i])
    print("Vendedor", i+1, total)

print("ntotal por dias")
for columna in range(3):
    total = 0
    for fila in range (2):
        total += ventas[fila][columna]
    print("Dia", columna + 1, total)
```
