## Introducción

En esta actividad voy a crear una tabla de Pitágoras utilizando Python. La tabla tendrá los resultados de las multiplicaciones del 1 al 10 y la voy a guardar dentro de una matriz, que es una lista que contiene otras listas.

Primero voy a construir la tabla utilizando dos ciclos `for`. Después voy a crear funciones para mostrarla y para buscar los resultados sin tener que volver a hacer las multiplicaciones.

También voy a agregar una parte donde el usuario pueda escribir dos números y el programa le muestre el resultado. Para evitar errores, voy a hacer que el programa revise que los números estén entre el 1 y el 10 antes de realizar la consulta.

## Parte A. Construcción de la tabla de Pitágoras

Primero voy a crear una matriz vacía donde voy a guardar todos los resultados de la tabla de Pitágoras.

Para hacerlo voy a utilizar dos ciclos `for`. El primero me sirve para recorrer los números del 1 al 10 y crear cada fila. El segundo me permite recorrer nuevamente los números del 1 al 10 para ir obteniendo los resultados de cada columna.

También voy a utilizar `append()` para ir agregando los resultados a cada fila y después guardar esa fila completa dentro de la matriz.

En esta primera parte sí utilizo el signo `*`, porque aquí necesito obtener los resultados para construir la tabla. Al terminar voy a tener una matriz con 10 filas y 10 columnas.

```python
# Creo una lista vacía para guardar toda mi tabla
tabla_pitagoras = []

# Recorro los números del 1 al 10 para crear las filas
for fila in range(1, 11):

    # Creo una lista vacía para guardar los números de cada fila
    fila_actual = []

    # Recorro los números del 1 al 10 para llenar las columnas
    for columna in range(1, 11):

        # Obtengo el resultado de cada multiplicación
        resultado = fila * columna

        # Agrego el resultado a la fila que estoy creando
        fila_actual.append(resultado)

    # Guardo la fila completa dentro de mi matriz
    tabla_pitagoras.append(fila_actual)

# Muestro la matriz para comprobar que se creó correctamente
print("Mi tabla de Pitágoras se creó correctamente.")
print("Cantidad de filas:", len(tabla_pitagoras))
print("Cantidad de columnas:", len(tabla_pitagoras[0]))
```

## Parte B y C. Funciones para mostrar y consultar la tabla

Después de crear mi matriz, voy a trabajar con dos funciones. Cada una tiene una tarea diferente.

### 1. Función imprimir_tabla()

Con esta función voy a mostrar toda mi tabla de Pitágoras en pantalla.

La función recibe la matriz mediante el parámetro `tabla` y utiliza dos ciclos `for` para recorrer sus filas y sus números.

Para que la tabla se vea más ordenada, utilizo `end="\t"`, que sirve para separar los números con espacios de tabulación. Al terminar cada fila utilizo `print()` para pasar a la siguiente línea.

En esta función no utilizo `return`, porque solamente necesito que muestre la información en pantalla y no que me regrese un resultado.

### 2. Función consultar_producto()

Esta función me sirve para buscar el resultado de una multiplicación dentro de mi matriz.

Le paso tres datos: la tabla, el renglón y la columna que quiero consultar.

Aquí hay algo importante: no voy a utilizar el signo `*` para calcular el resultado. En lugar de eso, voy a buscar directamente el número que ya está guardado en la matriz.

Utilizo `renglon - 1` y `columna - 1` porque Python comienza a contar las posiciones desde cero. Por ejemplo, si quiero consultar el renglón 7 y la columna 8, necesito acceder a las posiciones 6 y 7.

En esta función sí utilizo `return`, porque necesito que me regrese el resultado para poder guardarlo en una variable y mostrarlo después.

```python
# Esta función me permite mostrar toda mi tabla
def imprimir_tabla(tabla):

    print("TABLA DE PITÁGORAS")
    print("------------------------------")

    # Recorro cada fila de mi matriz
    for fila in tabla:

        # Recorro cada número que se encuentra en la fila
        for numero in fila:

            # Muestro cada número separado por una tabulación
            print(numero, end="\t")

        # Paso a la siguiente línea cuando termina cada fila
        print()


# Esta función me permite buscar un resultado dentro de mi matriz
def consultar_producto(tabla, renglon, columna):

    # Busco el resultado utilizando las posiciones de la matriz
    return tabla[renglon - 1][columna - 1]


# Muestro mi tabla completa para revisar que esté correcta
imprimir_tabla(tabla_pitagoras)
```

## Parte D. Programa principal

Ahora voy a crear la parte principal de mi programa. Aquí es donde el usuario podrá escribir los dos números que quiere consultar.

Primero voy a pedir el número del renglón y después el número de la columna. Para guardar esos datos voy a utilizar `input()` y `int()`.

También voy a utilizar un ciclo `while` para revisar que los dos números estén dentro del rango permitido, que es del 1 al 10.

Si alguien escribe un número menor que 1 o mayor que 10, el programa va a mostrar un mensaje indicando que el valor no es válido y le va a volver a pedir el dato.

Por último, cuando los dos números sean correctos, voy a llamar a mi función `consultar_producto()`. El resultado que me regrese lo voy a guardar en una variable llamada `resultado_final` y después lo voy a mostrar en pantalla.

De esta manera puedo consultar la tabla sin tener que hacer nuevamente la multiplicación.

```python
# Pido al usuario el número del renglón
renglon = int(input("Escribe el número del renglón (1 al 10): "))

# Reviso que el renglón esté dentro del rango permitido
while renglon < 1 or renglon > 10:

    print("El número no es válido. Debe estar entre 1 y 10.")

    renglon = int(input("Vuelve a escribir el número del renglón: "))


# Pido al usuario el número de la columna
columna = int(input("Escribe el número de la columna (1 al 10): "))

# Reviso que la columna esté dentro del rango permitido
while columna < 1 or columna > 10:

    print("El número no es válido. Debe estar entre 1 y 10.")

    columna = int(input("Vuelve a escribir el número de la columna: "))


# Consulto el resultado directamente en mi matriz
resultado_final = consultar_producto(tabla_pitagoras, renglon, columna)

# Muestro el resultado que encontré
print("--------------------------------")
print("RESULTADO DE MI CONSULTA")
print("--------------------------------")
print(renglon, "x", columna, "=", resultado_final)
```
