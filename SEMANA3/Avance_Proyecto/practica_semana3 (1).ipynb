## Ejercicio 2. Modelo Entrada–Proceso–Salida (EPS)

En este ejercicio voy a identificar los datos de entrada, el proceso y la salida de un problema de compra.

**Problema:** Una persona compra productos por un valor de $500 y recibe un descuento del 10 %. Quiero calcular cuánto dinero se descuenta y cuánto debe pagar al final.

**Entrada:**
- Subtotal de la compra: $500.
- Porcentaje de descuento: 10 %.

**Proceso:**
1. Multiplico el subtotal por el porcentaje de descuento.
2. Resto el descuento al subtotal.

**Salida esperada:**
- Descuento: $50.
- Total a pagar: $450.

```python
# Ejercicio 2: modelo Entrada-Proceso-Salida

subtotal = 500
porcentaje_descuento = 0.10

descuento = subtotal * porcentaje_descuento
total = subtotal - descuento

print("Subtotal: $", subtotal)
print("Descuento: $", descuento)
print("Total a pagar: $", total)
```

**Conclusión:** Con este ejercicio practiqué el modelo Entrada–Proceso–Salida. Identifiqué los datos que necesita el programa, las operaciones que debe realizar y los resultados que debe mostrar.

## Ejercicio 4. Estructuras condicionales

En este ejercicio voy a practicar cómo tomar decisiones en Python utilizando `if`, `elif` y `else`.

**Problema:** En una cafetería se ofrece un descuento del 10 % a los estudiantes cuando el subtotal de su compra es de $150 o más.

**Entrada:**
- El subtotal de la compra.
- Si el cliente es estudiante.

**Proceso:**
1. Verifico si el cliente es estudiante.
2. Compruebo si su compra es de $150 o más.
3. Si cumple ambas condiciones, calculo el descuento.
4. Si no cumple las condiciones, no aplico descuento.

**Salida esperada:** El programa muestra el subtotal, el descuento y el total después del descuento.

```python
# Ejercicio 4: estructuras condicionales

subtotal = float(input("Ingresa el subtotal de la compra: "))
respuesta = input("¿El cliente es estudiante? (S/N): ").strip().upper()

if respuesta not in ("S", "N"):
    print("Respuesta no válida. Escribe S o N.")
elif respuesta == "S" and subtotal >= 150:
    descuento = subtotal * 0.10
    total = subtotal - descuento
    print("Descuento aplicado: $", round(descuento, 2))
    print("Total después del descuento: $", round(total, 2))
else:
    descuento = 0
    total = subtotal
    print("No aplica descuento.")
    print("Total a pagar: $", round(total, 2))
```

```python
# Pruebas automáticas de las condiciones

pruebas = [
    (195, "S"),
    (115, "S"),
    (195, "N")
]

for subtotal, respuesta in pruebas:
    if respuesta == "S" and subtotal >= 150:
        descuento = subtotal * 0.10
    else:
        descuento = 0

    total = subtotal - descuento

    print("--------------------")
    print("Subtotal: $", subtotal)
    print("Estudiante:", respuesta)
    print("Descuento: $", round(descuento, 2))
    print("Total: $", round(total, 2))
```

**Conclusión:** Aprendí a utilizar condicionales para aplicar descuentos según las condiciones establecidas. Con los tres casos de prueba comprobé que el descuento solo se aplica a estudiantes con compras de $150 o más.

## Ejercicio 5. Ciclo while y validación de datos

En este ejercicio voy a practicar el ciclo `while` para validar la cantidad de productos que ingresa un usuario.

**Problema:** En una cafetería no se permiten cantidades negativas de productos. Si el usuario ingresa una cantidad negativa, el programa debe solicitarla nuevamente.

**Entrada:** La cantidad de cafés que desea comprar.

**Proceso:**
1. Solicito una cantidad.
2. Verifico si es negativa.
3. Mientras sea negativa, vuelvo a solicitar el dato.
4. Cuando la cantidad sea válida, muestro el resultado.

**Salida esperada:** El programa acepta una cantidad igual o mayor que cero y vuelve a pedir el dato si es negativo.

```python
# Ejercicio 5: ciclo while

cantidad = int(input("¿Cuántos cafés deseas comprar? "))

while cantidad < 0:
    print("Error: la cantidad no puede ser negativa.")
    cantidad = int(input("Ingresa una cantidad válida: "))

print("Cantidad registrada:", cantidad)
print("Costo de los cafés: $", cantidad * 35)
```

## Ejercicio 6. Acumuladores

En este ejercicio voy a practicar cómo sumar varias cantidades para obtener el total de ventas de una cafetería.

**Problema:** Necesito sumar los importes de varios pedidos para conocer las ventas acumuladas del día.

**Entrada:** Los importes de los pedidos.

**Proceso:** Sumar cada importe a un acumulador.

**Salida esperada:** El total de las ventas registradas.

```python
# Ejercicio 6: acumulador de ventas

total_dia = 0

pedido1 = 133.40
pedido2 = 203.58

total_dia = total_dia + pedido1
total_dia = total_dia + pedido2

print("Venta del pedido 1: $", pedido1)
print("Venta del pedido 2: $", pedido2)
print("Ventas acumuladas: $", round(total_dia, 2))
```

**Conclusión:** Aprendí a utilizar un acumulador para sumar los importes de varios pedidos y calcular las ventas totales del día.

## Ejercicio 7. Depuración de programas con PDB

En este ejercicio voy a practicar cómo revisar la ejecución de un programa en Python utilizando la herramienta de depuración `pdb`.

**Problema:** En una cafetería necesito comprobar que el cálculo del subtotal de una compra sea correcto. Para hacerlo, voy a pausar el programa y revisar los valores de sus variables durante la ejecución.

**Entrada:**

- Precio de cada café: $35.
- Cantidad de cafés: 3.

**Proceso:**

1. Asigno los valores del precio y la cantidad a sus variables.
2. Utilizo `pdb` para detener temporalmente la ejecución.
3. Reviso los valores de las variables con los comandos de depuración.
4. Continúo la ejecución y compruebo el cálculo del subtotal.

**Salida esperada:** El programa muestra un precio de $35, una cantidad de 3 cafés y un subtotal de $105.

```python
# Ejercicio 7: Depuración de programas con PDB

import pdb

precio = 35
cantidad = 3

pdb.set_trace()

subtotal = precio * cantidad

print("Precio por café: $", precio)
print("Cantidad de cafés:", cantidad)
print("Subtotal de la compra: $", subtotal)
```

**Conclusión:** Aprendí a utilizar PDB para revisar las variables paso a paso y comprobar que el cálculo del subtotal fuera correcto. Esto me ayuda a detectar errores durante la ejecución de un programa.

## Ejercicio 8. Ciclos for

En este ejercicio voy a practicar el ciclo `for` para repetir instrucciones una cantidad determinada de veces.

**Problema:** En una cafetería quiero contar cuántos números pares hay del 1 al 10 para practicar el uso de ciclos y condicionales.

**Entrada:**

- Números enteros del 1 al 10.

**Proceso:**

1. Utilizo un ciclo `for` para recorrer los números del 1 al 10.
2. Verifico cuáles son divisibles entre 2.
3. Cada vez que encuentro un número par, aumento un contador.
4. Muestro los números pares encontrados y la cantidad total.

**Salida esperada:** Los números pares son 2, 4, 6, 8 y 10. En total hay 5 números pares.

```python
# Ejercicio 8: ciclo for y contador

contador_pares = 0

print("Números pares del 1 al 10:")

for numero in range(1, 11):
    if numero % 2 == 0:
        print(numero)
        contador_pares += 1

print("Cantidad total de números pares:", contador_pares)
```

**Conclusión:** Aprendí a utilizar el ciclo for para recorrer una secuencia de números y un contador para registrar cuántos cumplen una condición.
