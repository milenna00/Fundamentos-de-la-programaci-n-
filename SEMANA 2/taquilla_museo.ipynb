## Descripción

Este programa simula una taquilla de museo. Calcula el precio de entrada según la edad, aplica un solo descuento cuando corresponde y valida los datos ingresados.

La taquilla permite atender varios visitantes y, al escribir `fin`, muestra un corte de caja con el número de visitantes cobrados, los descuentos y el total a pagar.

```python
# Actividad 2 - La taquilla del museo
# Fundamentos de programacion
# Semana 2

# Precios de los boletos
PRECIO_ADULTO = 45
PRECIO_NINO = 30
PRECIO_MENOR = 0

# Descuentos
DESCUENTO_ADULTO = 0.12
DESCUENTO_ESTUDIANTE = 0.10
DESCUENTO_PROFESOR = 0.10
DESCUENTO_MENOR = 0.0


# Funcion para calcular el precio de un visitante
def calcular_boleto(edad, opcion=None):

    # Revisar que la edad sea valida
    if edad < 0 or edad > 110:
        return None

    # Valores iniciales
    precio_base = 0
    descuento = 0.0
    tipo = "Ninguno"

    # Los menores de 3 años entran gratis
    if edad < 3:
        precio_base = PRECIO_MENOR
        descuento = DESCUENTO_MENOR
        tipo = "Menor de 3 años"

    # Los niños de 3 a 17 años pagan 30 pesos
    elif edad <= 17:
        precio_base = PRECIO_NINO
        descuento = DESCUENTO_MENOR
        tipo = "Menor de edad"

    # Adultos mayores de 60 años tienen 12% automatico
    elif edad >= 60:
        precio_base = PRECIO_ADULTO
        descuento = DESCUENTO_ADULTO
        tipo = "Adulto mayor"

    # Adultos de 18 a 59 años
    else:
        precio_base = PRECIO_ADULTO

        if opcion == "1":
            tipo = "Estudiante"
            descuento = DESCUENTO_ESTUDIANTE

        elif opcion == "2":
            tipo = "Profesor"
            descuento = DESCUENTO_PROFESOR

        elif opcion == "3":
            tipo = "Ninguno"
            descuento = 0.0

        else:
            return None

    # Calcular el descuento en pesos
    cantidad_descuento = round(precio_base * descuento, 2)

    # Calcular lo que paga el visitante
    costo = round(precio_base - cantidad_descuento, 2)

    return {
        "edad": edad,
        "tipo": tipo,
        "precio_base": precio_base,
        "descuento": cantidad_descuento,
        "costo": costo
    }


# Funcion principal de la taquilla
def taquilla(entradas=None):

    # Contadores y acumuladores
    contador_visitantes = 0
    total_descuentos = 0.0
    total_pagar = 0.0

    # Esta variable sirve para las pruebas automaticas
    posicion = 0

    print("\n=== TAQUILLA DEL MUSEO ===")

    # El ciclo sigue hasta escribir fin
    while True:

        mensaje = "\nEdad del visitante (o 'fin' para cerrar): "

        # Si no hay pruebas, pedir datos al usuario
        if entradas is None:
            entrada = input(mensaje).strip()

        # Si estamos probando, usar los datos de la lista
        else:
            if posicion >= len(entradas):
                raise ValueError("La prueba no tiene la entrada fin.")

            entrada = str(entradas[posicion]).strip()
            posicion += 1
            print(mensaje + entrada)

        # Revisar si quieren cerrar la taquilla
        if entrada.lower() == "fin":
            print("\nLa taquilla ha cerrado.")
            break

        # Convertir la edad a numero entero
        try:
            edad = int(entrada)

        except ValueError:
            print("Edad invalida. Debes escribir un numero entero.")
            continue

        # No aceptar edades negativas ni mayores de 110
        if edad < 0 or edad > 110:
            print("Edad invalida. No se cobra este visitante.")
            continue

        # Por defecto no se necesita preguntar el tipo
        opcion = None

        # Solo preguntar a personas de 18 a 59 años
        if edad >= 18 and edad < 60:

            print("\nTipo de visitante:")
            print("1. Estudiante")
            print("2. Profesor")
            print("3. Ninguno")

            # Repetir hasta recibir una opcion correcta
            while True:

                mensaje_tipo = "Introduce una opcion del 1 al 3: "

                if entradas is None:
                    opcion = input(mensaje_tipo).strip()

                else:
                    if posicion >= len(entradas):
                        raise ValueError(
                            "Falta una opcion en los datos de prueba."
                        )

                    opcion = str(entradas[posicion]).strip()
                    posicion += 1
                    print(mensaje_tipo + opcion)

                # Revisar las opciones del menu
                if opcion in ("1", "2", "3"):
                    break

                else:
                    print("Opcion invalida. Intenta nuevamente.")

        # Calcular el boleto del visitante
        boleto = calcular_boleto(edad, opcion)

        # Sumar al contador y los acumuladores
        contador_visitantes += 1
        total_descuentos += boleto["descuento"]
        total_pagar += boleto["costo"]

        # Mostrar el detalle del visitante
        print(
            f"Visitante {contador_visitantes} | "
            f"Edad: {boleto['edad']} | "
            f"Tipo: {boleto['tipo']} | "
            f"Base: ${boleto['precio_base']:.2f} | "
            f"Descuento: ${boleto['descuento']:.2f} | "
            f"A pagar: ${boleto['costo']:.2f}"
        )

    # Mostrar el corte de caja
    print("\n===================================")
    print("=== CORTE DE CAJA ===")
    print("===================================")

    print(f"Visitantes cobrados: {contador_visitantes}")
    print(f"Total en descuentos: ${total_descuentos:.2f}")
    print(f"TOTAL A PAGAR: ${total_pagar:.2f}")

    print("===================================")

    # Regresar los resultados para comprobar las pruebas
    return {
        "visitantes": contador_visitantes,
        "descuentos": round(total_descuentos, 2),
        "total": round(total_pagar, 2)
    }


# Funcion para ejecutar un caso de prueba
def probar_caso(numero, entradas, esperado, visitantes, descuentos):

    print("\n" + "=" * 45)
    print(f"CASO DE PRUEBA {numero}")
    print("=" * 45)

    # Ejecutar el programa con las entradas de prueba
    resultado = taquilla(entradas)

    # Comparar los resultados
    coincide_total = abs(resultado["total"] - esperado) < 0.001

    coincide_visitantes = resultado["visitantes"] == visitantes

    coincide_descuentos = (
        abs(resultado["descuentos"] - descuentos) < 0.001
    )

    print("\n=== RESULTADO DE LA PRUEBA ===")
    print(f"Total esperado: ${esperado:.2f}")
    print(f"Total obtenido: ${resultado['total']:.2f}")

    print(f"Visitantes esperados: {visitantes}")
    print(f"Visitantes obtenidos: {resultado['visitantes']}")

    print(f"Descuentos esperados: ${descuentos:.2f}")
    print(f"Descuentos obtenidos: ${resultado['descuentos']:.2f}")

    # Revisar si todos los resultados son correctos
    if coincide_total and coincide_visitantes and coincide_descuentos:
        print("Resultado: CORRECTO. Los valores coinciden.")
        return True

    else:
        print("Resultado: ERROR. Los valores no coinciden.")
        return False


# Funcion para ejecutar todas las pruebas
def ejecutar_pruebas():

    print("\n=== CASOS DE PRUEBA ===")

    resultados = []

    # Caso 1: diferentes edades y descuentos
    # 25 años profesor, 2 años, 15 años y 67 años
    resultados.append(
        probar_caso(
            1,
            ["25", "2", "2", "15", "67", "fin"],
            110.10,
            4,
            9.90
        )
    )

    # Caso 2: edad negativa
    # La edad -5 no debe cobrarse
    # La persona de 30 años paga 45 pesos
    resultados.append(
        probar_caso(
            2,
            ["-5", "30", "3", "fin"],
            45.00,
            1,
            0.00
        )
    )

    # Caso 3: cerrar sin registrar visitantes
    resultados.append(
        probar_caso(
            3,
            ["fin"],
            0.00,
            0,
            0.00
        )
    )

    # Caso 4: prueba propia de descuento unico
    # Dos adultos mayores, estudiante y profesor
    resultados.append(
        probar_caso(
            4,
            ["65", "60", "25", "1", "35", "2", "fin"],
            160.20,
            4,
            19.80
        )
    )

    # Prueba adicional: opciones incorrectas
    # Primero se escribe una opcion invalida
    resultados.append(
        probar_caso(
            6,
            ["25", "9", "abc", "2", "fin"],
            40.50,
            1,
            4.50
        )
    )

    # Resumen de todas las pruebas
    print("\n===================================")
    print("=== RESUMEN DE PRUEBAS ===")
    print("===================================")

    correctas = sum(resultados)
    total_pruebas = len(resultados)

    print(f"Pruebas realizadas: {total_pruebas}")
    print(f"Pruebas correctas: {correctas}")
    print(f"Pruebas incorrectas: {total_pruebas - correctas}")

    if correctas == total_pruebas:
        print("Todas las pruebas pasaron correctamente.")
    else:
        print("Hay pruebas que necesitan revision.")

```

**Cálculos esperados:**

```python
resultado4 = taquilla(["65", "60", "25", "1", "35", "2", "fin"])
```

```python
# COMPARACIÓN DE RESULTADOS ESPERADOS Y OBTENIDOS

casos = [
    ("Caso 1", resultado1, 4, 9.90, 110.10),
    ("Caso 2", resultado2, 1, 0.00, 45.00),
    ("Caso 3", resultado3, 0, 0.00, 0.00),
    ("Caso 4", resultado4, 4, 19.80, 160.20)
]

pruebas_correctas = 0

for nombre, obtenido, visitantes, descuentos, total in casos:

    coincide_visitantes = (
        obtenido["visitantes"] == visitantes
    )

    coincide_descuentos = (
        abs(obtenido["descuentos"] - descuentos) < 0.001
    )

    coincide_total = (
        abs(obtenido["total"] - total) < 0.001
    )

    coincide = (
        coincide_visitantes
        and coincide_descuentos
        and coincide_total
    )

    print("\n" + nombre)

    print(
        f"Visitantes: esperado {visitantes}, "
        f"obtenido {obtenido['visitantes']}"
    )

    print(
        f"Descuentos: esperado ${descuentos:.2f}, "
        f"obtenido ${obtenido['descuentos']:.2f}"
    )

    print(
        f"Total a pagar: esperado ${total:.2f}, "
        f"obtenido ${obtenido['total']:.2f}"
    )

    if coincide:
        print("Resultado: CORRECTO")
        pruebas_correctas += 1
    else:
        print("Resultado: INCORRECTO")

print("\n=== RESUMEN DE PRUEBAS ===")
print(f"Pruebas correctas: {pruebas_correctas} de {len(casos)}")
```
