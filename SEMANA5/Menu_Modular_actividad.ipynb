## Introducción

En esta actividad voy a desarrollar un programa en Python que funciona mediante un menú interactivo. Mi objetivo es practicar los temas de esta semana: tuplas, diccionarios, excepciones y cadenas de texto o strings.

Voy a organizar el programa utilizando funciones, para que cada opción del menú realice una tarea diferente. También utilizaré un ciclo `while` para que el menú se repita hasta que decida salir.

Al terminar, realizaré diez pruebas para comprobar que el programa funcione correctamente, tanto cuando ingreso datos válidos como cuando ingreso información incorrecta.

```python
# ==========================================
# PARTE A. TRABAJO CON TUPLAS
# ==========================================

# Funcion para sumar los elementos de una tupla

def calcular_total(datos):
    total = 0

    for valor in datos:
        total += valor

    return total


# Funcion principal de la seccion de tuplas

def seccion_tuplas():

    numeros = (8, 3, 15, 6, 10)

    print("\n--- TRABAJO CON TUPLAS ---")

    # Mostrar la tupla original
    print("Tupla original:", numeros)

    # Mostrar el tercer elemento
    print("El tercer elemento es:", numeros[2])

    # Solicitar dos numeros al usuario
    primero = int(input("Ingresa un primer numero: "))
    segundo = int(input("Ingresa un segundo numero: "))

    # Crear una nueva tupla sin modificar la original
    nueva_tupla = numeros + (primero, segundo)

    print("Nueva tupla:", nueva_tupla)

    # Convertir la nueva tupla en lista y ordenarla
    lista_ordenada = list(nueva_tupla)
    lista_ordenada.sort()

    print("Lista ordenada:", lista_ordenada)

    # Calcular la suma de la nueva tupla
    resultado = calcular_total(nueva_tupla)

    print("La suma de la nueva tupla es:", resultado)
```

## Parte A. Tuplas

En esta sección utilicé una tupla llamada `numeros`, que contiene cinco valores. Primero mostré su tercer elemento utilizando `numeros[2]`, porque en Python las posiciones comienzan desde cero.

Después solicité dos números al usuario y los agregué a una nueva tupla. De esta manera conservé la tupla original sin modificarla.

Posteriormente convertí la nueva tupla en una lista utilizando `list()` y ordené sus elementos con `.sort()`.

Finalmente utilicé la función `calcular_total()`, que recorre los números mediante un ciclo `for`, los suma y regresa el resultado con `return`.

```python
# ==========================================
# PARTE B. TRABAJO CON DICCIONARIOS
# ==========================================

# Funcion para buscar el telefono de un contacto

def buscar_telefono(contactos, nombre):
    return contactos.get(nombre)


# Funcion principal de la seccion de diccionarios

def seccion_diccionarios():

    contactos = {
        "Regina": "4421234567",
        "Viridiana": "4429876543",
        "Mariana": "4425678910"
    }

    print("\n--- TRABAJO CON DICCIONARIOS ---")

    # Solicitar un nuevo contacto
    nombre_nuevo = input("Ingresa el nombre del nuevo contacto: ")
    telefono_nuevo = input("Ingresa su numero de telefono: ")

    # Agregar el contacto al diccionario
    contactos[nombre_nuevo] = telefono_nuevo

    print("\nContacto agregado correctamente.")

    # Mostrar solamente los nombres
    print("\nNombres de los contactos registrados:")

    for nombre in contactos:
        print(nombre)

    # Solicitar el nombre que se desea buscar
    nombre_buscar = input("\nEscribe el nombre que deseas buscar: ")

    telefono_encontrado = buscar_telefono(contactos, nombre_buscar)

    # Comprobar si el contacto existe
    if telefono_encontrado is not None:
        print("El telefono de", nombre_buscar, "es:", telefono_encontrado)

    else:
        print(nombre_buscar, "no esta en tus contactos.")
```

## Parte B. Diccionarios

En esta sección utilicé un diccionario llamado `contactos`, donde guardé nombres como claves y números telefónicos como valores.

Primero registré tres contactos y después solicité los datos de una persona nueva para agregarla al diccionario.

Utilicé un ciclo `for` para recorrer las claves y mostrar únicamente los nombres registrados.

También creé la función `buscar_telefono()`, que utiliza `.get()` para buscar el teléfono de una persona. Este método permite que, si el nombre no existe, el programa reciba `None` en lugar de generar un error.

Finalmente utilicé una condición `if` y `else` para mostrar el teléfono cuando el contacto existe o un mensaje cuando no está registrado.

```python
# ==========================================
# PARTE C. MANEJO DE EXCEPCIONES
# ==========================================

def seccion_excepciones():

    print("\n--- MANEJO DE EXCEPCIONES ---")

    try:

        # Solicitar dos numeros enteros
        numero_uno = int(input("Ingresa el primer numero entero: "))
        numero_dos = int(input("Ingresa el segundo numero entero: "))

        # Calcular y mostrar la suma
        suma = numero_uno + numero_dos

        print("La suma de los numeros es:", suma)

        # Calcular y mostrar la division
        division = numero_uno / numero_dos

        print("La division del primer numero entre el segundo es:", division)

    except ValueError:

        print("Error: debes ingresar numeros enteros y no dejar campos vacios.")

    except ZeroDivisionError:

        print("No se puede dividir entre cero. Intenta con otro numero.")
```

## Parte C. Manejo de excepciones

En esta sección utilicé `try` y `except` para controlar posibles errores al solicitar dos números enteros.

Primero solicité los valores y los convertí con `int()`. Si los datos son correctos, el programa muestra su suma y su división.

Utilicé `except ValueError` para controlar cuando ingreso letras o dejo un campo vacío, ya que esos valores no se pueden convertir correctamente a números enteros.

También utilicé `except ZeroDivisionError` para controlar específicamente el caso en el que intento dividir entre cero.

Con esto evito que el programa se detenga inesperadamente y puedo mostrar mensajes que explican qué sucedió.

```python
# ==========================================
# PARTE D. TRABAJO CON STRINGS
# ==========================================

# Funcion para contar las palabras de un texto

def contar_palabras(texto):

    palabras = texto.split()

    return len(palabras)


# Funcion principal de la seccion de strings

def seccion_strings():

    mensaje = "Estoy aprendiendo programacion en Python"

    print("\n--- TRABAJO CON STRINGS ---")

    # Mostrar el mensaje original
    print("Mensaje original:", mensaje)

    # Mostrar la longitud del mensaje
    print("Longitud del mensaje:", len(mensaje))

    # Convertir el mensaje a mayusculas
    print("Mensaje en mayusculas:", mensaje.upper())

    # Reemplazar una palabra
    mensaje_nuevo = mensaje.replace("programacion", "tecnologia")

    print("Mensaje con la palabra reemplazada:", mensaje_nuevo)

    # Contar las palabras del mensaje
    cantidad = contar_palabras(mensaje)

    print("Cantidad de palabras:", cantidad)
```

## Parte D. Strings

En esta sección trabajé con una cadena de texto llamada `mensaje`, que contiene más de cuatro palabras.

Utilicé `len()` para conocer la cantidad de caracteres, incluyendo los espacios. Después utilicé `.upper()` para convertir el mensaje a mayúsculas.

También utilicé `.replace()` para cambiar la palabra "programacion" por "tecnologia".

Finalmente creé la función `contar_palabras()`, que utiliza `.split()` para separar el texto en palabras y `len()` para contarlas. La función regresa el resultado mediante `return`.

```python
# ==========================================
# PARTE E. MENU PRINCIPAL
# ==========================================

def menu_principal():

    opcion = ""

    while opcion != "5":

        print("\n================================")
        print("          MENU PRINCIPAL")
        print("================================")
        print("1. Tuplas")
        print("2. Diccionarios")
        print("3. Excepciones")
        print("4. Strings")
        print("5. Salir")
        print("================================")

        opcion = input("Elige una opcion: ")

        if opcion == "1":

            seccion_tuplas()

        elif opcion == "2":

            seccion_diccionarios()

        elif opcion == "3":

            seccion_excepciones()

        elif opcion == "4":

            seccion_strings()

        elif opcion == "5":

            print("\nEl programa ha terminado. Gracias por utilizarlo.")

        else:

            print("\nOpcion invalida. Selecciona una opcion del 1 al 5.")


# Ejecutar el menu principal

menu_principal()
```

## Parte E. Menú principal

Para integrar todas las secciones creé la función `menu_principal()`.

Utilicé un ciclo `while` para que las opciones aparezcan continuamente hasta que seleccione la opción 5.

Cada opción llama a una función diferente:

- **Opción 1:** ejecuta `seccion_tuplas()`.
- **Opción 2:** ejecuta `seccion_diccionarios()`.
- **Opción 3:** ejecuta `seccion_excepciones()`.
- **Opción 4:** ejecuta `seccion_strings()`.
- **Opción 5:** termina el programa.

También utilicé `else` para mostrar un mensaje cuando ingreso una opción que no existe. El menú vuelve a aparecer después de ese mensaje.

De esta manera, cada sección está organizada por separado y el menú se encarga de controlar cuál función debe ejecutarse.
