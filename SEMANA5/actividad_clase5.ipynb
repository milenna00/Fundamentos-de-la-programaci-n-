```python
precios = {"playera": 350, "gorra": 220, "parche": 80, "sudadera": 650}
carrito = {"playera": 2, "parche": 3}

total = 0
for producto, cantidad in carrito.items():
    total += precios[producto] * cantidad

print(f"Total: ${total}")
```

```python
operaciones = {
    lambda: int(""),
    lambda: 10 / 0,
    lambda: [1, 2][5],
    lambda: {"a": 1}["b"],
    lambda: "a" + 1
}

for operacion in operaciones:
    try:
        operacion()
    except Exception as e:
        print(type(e).__name__)
```
