## Ejercicio 1 — Bitácora de la semana

```python
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(km):
    total = 0
    dias_rodados = 0
    indice_mayor = 0
    mayor = km[0]
    for i in range(len(km)):
        total += km[i]
        if km[i] > 0:
            dias_rodados += 1
        if km[i] > mayor:
            mayor = km[i]
            indice_mayor = i
    promedio = round(total / dias_rodados, 1) if dias_rodados > 0 else 0
    return (total, dias_rodados, promedio, indice_mayor)

total, dias_rodados, promedio, indice = resumen(km)
print(f"Total: {total} km")
print(f"Días con rodada: {dias_rodados}")
print(f"Promedio por día rodado: {promedio} km")
print(f"Día más largo: {dias[indice]} ({km[indice]} km)")
```

## Ejercicio 2 — La votación del festival

```python
votos = [("Metallica", 48), ("Slayer", 22), ("Megadeth", 35), ("Anthrax", 15), ("Iron Maiden", 40)]

def promedio_votos(lista):
    suma = 0
    for banda, cantidad in lista:
        suma += cantidad
    return suma / len(lista)

ganadora = votos[0]
for banda, cantidad in votos:
    if cantidad > ganadora[1]:
        ganadora = (banda, cantidad)

promedio = promedio_votos(votos)
print("Ganador:", ganadora[0], "con", ganadora[1], "votos")
print("Promedio:", promedio)
print("Arriba del promedio:")
for banda, cantidad in votos:
    if cantidad > promedio:
        print("-", banda, "(", cantidad, ")")
```

## Ejercicio 3 — Tramos de la ruta a la Sierra

```python
paradas = (("Querétaro", 0), ("San Juan del Río", 52), ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))

mayor = 0
inicio_mayor = ""
fin_mayor = ""

for i in range(len(paradas) - 1):
    nombre1, km1 = paradas[i]
    nombre2, km2 = paradas[i + 1]
    distancia = km2 - km1
    print(nombre1, "→", nombre2 + ":", distancia, "km")
    if distancia > mayor:
        mayor = distancia
        inicio_mayor = nombre1
        fin_mayor = nombre2

print("Tramo más largo:", inicio_mayor, "→", fin_mayor, "(", mayor, "km)")
```

## Ejercicio 4 — ¿Con chamarra o sin chamarra?

```python
temps = (8, 14, 22, 27, 31, 19, 11)

def clasificar(t):
    if t < 12:
        return "frío"
    elif t <= 24:
        return "templado"
    else:
        return "calor"

etiquetas = []
contador = {"frío": 0, "templado": 0, "calor": 0}

for temperatura in temps:
    etiqueta = clasificar(temperatura)
    etiquetas.append(etiqueta)
    contador[etiqueta] += 1

predomina = ""
for etiqueta, cantidad in contador.items():
    if predomina == "" or cantidad > contador[predomina]:
        predomina = etiqueta

print(etiquetas)
print(contador)
print("Predomina:", predomina)
```

## Ejercicio 5 — ¿Cuánto dura el show?

```python
setlist = (("Intro", 95), ("Riff asesino", 245), ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg):
    minutos = seg // 60
    segundos = seg % 60
    if segundos < 10:
        return f"{minutos}:0{segundos}"
    return f"{minutos}:{segundos}"

total = 0
for cancion, segundos in setlist:
    print(cancion + ":", formato(segundos))
    total += segundos

print("Duración total:", formato(total))
```

## Ejercicio 7 — Boleta del grupo

```python
notas = {"Ana": [90, 85, 78], "Luis": [60, 72, 65], "Mía": [100, 95, 98], "Beto": [70, 68, 74]}

def promedio(lista):
    suma = 0
    for nota in lista:
        suma += nota
    return round(suma / len(lista), 1)

promedios = {}
for alumno, calificaciones in notas.items():
    promedios[alumno] = promedio(calificaciones)

for alumno, calificacion in promedios.items():
    if calificacion >= 70:
        estado = "Aprobado"
    else:
        estado = "Reprobado"
    print(alumno + ":", calificacion, "-", estado)

mejor = ""
for alumno, calificacion in promedios.items():
    if mejor == "" or calificacion > promedios[mejor]:
        mejor = alumno

print("Mejor promedio:", mejor)
```

## Ejercicio 10 — Traductor de mensajes de WhatsApp

```python
abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    palabras = mensaje.lower().split()
    nuevas = []
    for palabra in palabras:
        nuevas.append(abrev.get(palabra, palabra))
    return " ".join(nuevas)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

## Ejercicio 13 — Capturas sucias

`" 6 "` sí se convierte porque `int()` ignora los espacios al inicio y al final; `"8.5"` no se convierte porque tiene un punto decimal y `int()` requiere un número entero.

```python
capturas = ["10", "7", "x", "", "9", "8.5", " 6 "]

validas = []
invalidas = 0

for captura in capturas:
    try:
        numero = int(captura)
        validas.append(numero)
    except ValueError:
        invalidas += 1

promedio = round(sum(validas) / len(validas), 2)
print("Válidas:", validas)
print("Inválidas:", invalidas)
print("Promedio:", promedio)
```

## Ejercicio 15 — Cupones de descuento

```python
cupones = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}

def aplicar(total, codigo):
    codigo = codigo.strip().upper()
    try:
        porcentaje = cupones[codigo]
        return total - (total * porcentaje / 100)
    except KeyError:
        print("Cupón no válido")
        return total

for codigo in ["  moto20 ", "GRATIS", "metal10"]:
    resultado = aplicar(1200, codigo)
    print(repr(codigo), "→ $" + str(resultado))
```

## Ejercicio 16 — ¿Tu contraseña aguanta?

```python
def evaluar():
    fallas = []
    tiene_numero = False
    tiene_mayuscula = False
    tiene_espacio = False

    if len() < 8:
        fallas("mínimo 8 caracteres")

    for caracter in :
        if caracter.isdigit():
            tiene_numero = True
        if caracter.isupper():
            tiene_mayuscula = True
        if caracter == " ":
            tiene_espacio = True

    if not tiene_numero:
        fallas("falta un número")
    if not tiene_mayuscula:
        fallas("falta una mayúscula")
    if tiene_espacio:
        fallas("sin espacios")

    return fallas

for clave in ["metal", "Harley2008", "road glide 08", "KLR650moto"]:
    resultado = evaluar(clave)
    if len(resultado) == 0:
        print(clave + ": segura")
    else:
        print(clave + ":", resultado)
```

## Ejercicio 18 — Contador de vocales

```python
frase = "el rock nunca va a morir en esta ciudad"

def vocales(palabra):
    cantidad = 0
    for letra in palabra:
        if letra in "aeiou":
            cantidad += 1
    return cantidad

palabras = frase.split()
conteo = {}
for palabra in palabras:
    conteo[palabra] = vocales(palabra)

print(conteo)
mayor = ""
for palabra, cantidad in conteo.items():
    if mayor == "" or cantidad > conteo[mayor]:
        mayor = palabra
print("Más vocales:", mayor, "(" + str(conteo[mayor]) + ")")
```

## Ejercicio 19 — Usuarios del laboratorio

```python
alumnos = ["Bruce Dickinson", "Brian Dickinson", "Ana Lopez", "Andrea Lopez", "Angel Lopez"]

usados = []

def usuario(nombre):
    partes = nombre.lower().split()
    base = partes[0][0] + partes[1]
    candidato = base
    numero = 2
    while candidato in usados:
        candidato = base + str(numero)
        numero += 1
    usados.append(candidato)
    return candidato

for alumno in alumnos:
    print(alumno + " →", usuario(alumno))
```

```python
autos = [("ABC123", 9, 11) ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]

def cobro(entrada, salida):
    horas = salida - entrada
    if horas= 1:
        return 20
    return 20 + (horas - 1) * 15

cobros = {}
for placa, entrada, salida in autos:
    horas = salida - entrada
    cobros[placa] = cobro(entrada, salida)
    print(placa + ":", horas)

total = 0
for cantidad in cobros.values():
    total + cantidad
print("Total del día: $" +(total))
```

## Ejercicio 22 — Tabla del torneo

```python
partidos = (("Pumas", "Tigres", 2, 1), ("América", "Pumas", 0, 0), ("Tigres", "América", 3, 2),
            ("Pumas", "América", 1, 2), ("Tigres", "Pumas", 1, 1))

puntos = {}
for local, visitante, goles_local, goles_visitante in partidos:
    if local not in puntos:
        puntos[local] = 0
    if visitante not in puntos:
        puntos[visitante] = 0

    if goles_local > goles_visitante:
        puntos[local] += 3
    elif goles_local < goles_visitante:
        puntos[visitante] += 3
    else:
        puntos[local] += 1
        puntos[visitante] += 1

print(puntos)
campeon = ""
for equipo, cantidad in puntos.items():
    if campeon == "" or cantidad > puntos[campeon]:
        campeon = equipo
print("Campeón:", campeon, "con", puntos[campeon], "puntos")
```
