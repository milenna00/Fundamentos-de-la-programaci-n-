```python
n = int(input("ingresa un numero entero positivo: "))
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print(f"El factorial de {n} es {factorial}")
```

```python
n = int(input("ingresa un numero entero positivo: "))
suma = 0 
for i in range(1, n + 1):
    suma += i
    print(f"La suma de los primeros {n} numeros es: {suma}")
```
